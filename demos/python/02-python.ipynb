```python
import time
from functools import wraps

#*args：接收任意数量的位置参数
#**kwargs：接收任意数量的关键字参数
def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)#这行是整个装饰器真正执行原函数的地方
        print(f"耗时: {time.time() - start:.4f}s")
        return result
    return wrapper

@timer
def process_news(count):
    time.sleep(1)
    return count

process_news(10)
```

```python
def read_news(news_list):
    for news in news_list:
        yield news

news = ["新闻A", "新闻B", "新闻C"]

for item in read_news(news):
    print(item)
```

```python
#迭代器
it = iter([10, 20, 30])

print(next(it))  # 10
print(next(it))  # 20
print(next(it))  # 30
print(next(it))  # 抛出 StopIteration
```

```python
#生成器
def get_numbers():
    yield 1
    yield 2
    yield 3

gen = get_numbers()

print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3

# 此时并没有执行函数体，而是创建了一个生成器对象。
# 每次调用 next()，函数才会执行到下一个 yield，并返回对应的数据。
```
