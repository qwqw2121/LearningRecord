```python
#同步执行：
import time

def task(name):
    time.sleep(2)
    print(name)

task("A")
task("B")
```

```python
#异步执行：
import asyncio

async def task(name):
    await asyncio.sleep(2)
    print(name)

async def main():
    await asyncio.gather(
        task("A"),
        task("B")
    )

asyncio.run(main())
```

Jupyter 内核本身已经运行着一个事件循环。

直接改用 await：不要用 asyncio.run() 包裹协程，直接在单元格中 await 你的异步函数。Jupyter 支持顶层 await。

```python
import asyncio

async def fetch_news():
    print("开始请求")
    await asyncio.sleep(2)
    print("请求完成")
    return "新闻数据"

async def main():
    result = await fetch_news()
    print(result)

asyncio.run(main())
```
