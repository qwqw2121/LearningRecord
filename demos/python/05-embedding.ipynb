```python
import numpy as np

a = np.array([1, -2, 3])
b = np.array([2, -3, 4])

# 余弦相似度
cosine = np.dot(a, b) / (
    np.linalg.norm(a) * np.linalg.norm(b)
)

# 欧氏距离
euclidean = np.linalg.norm(a - b)

# 内积
dot_product = np.dot(a, b)

print("余弦相似度:", cosine)
print("欧氏距离:", euclidean)
print("内积:", dot_product)
```
